# Leitor dos consolidados — Projeto Integrador III

Este notebook baixa todos os arquivos Markdown da pasta `consolidados/` do GitHub para o armazenamento temporário do Colab (`/content/consolidados/`) e também os exibe nesta página. Para carregar os documentos, use **Runtime → Run all**. Não é necessário instalar pacotes nem enviar arquivos manualmente.

É necessária conexão com a internet e acesso ao repositório no GitHub.

```python
from urllib.request import Request, urlopen
from IPython.display import Markdown, display
from pathlib import Path

BASE_URL = "https://raw.githubusercontent.com/Megalonnix/proj_ivan_gustavo_jorge_PI3/main/consolidados"
PASTA_DESTINO = Path("/content/consolidados")
PASTA_DESTINO.mkdir(parents=True, exist_ok=True)
ARQUIVOS = [
    "00_FICHA_PROJETO.md",
    "aula00_consolidado.md",
    "aula00_parteD_consolidado.md",
    "aula01_consolidado.md",
    "aula01_parteD_consolidado.md",
    "aula02_consolidado.md",
    "aula02_parteD_consolidado.md",
    "aula03_consolidado.md",
    "aula03_parteD_consolidado.md",
    "aula04_consolidado.md",
    "aula04_parteD_consolidado.md",
    "aula05b_consolidado.md",
    "aula05b_parteD_consolidado.md",
    "aula06_consolidado.md",
    "aula06_parteD_consolidado.md",
    "aula07_consolidado.md",
    "aula07_parteD_consolidado.md",
]

for nome in ARQUIVOS:
    requisicao = Request(f"{BASE_URL}/{nome}", headers={"User-Agent": "Mozilla/5.0"})
    with urlopen(requisicao, timeout=30) as resposta:
        conteudo = resposta.read().decode("utf-8")
    (PASTA_DESTINO / nome).write_text(conteudo, encoding="utf-8")
    display(Markdown(f"# {nome}\n\n{conteudo}"))

print(f"\n{len(ARQUIVOS)} arquivos salvos em: {PASTA_DESTINO}")
```
