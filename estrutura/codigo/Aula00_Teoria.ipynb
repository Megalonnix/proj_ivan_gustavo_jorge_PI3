## Módulo 1 — Rodar, atribuir, e o corpus como vetor
Criação do corpus canônico `docs`, indexação em base 1, vetor vazio `docs[0]` e vetorização.

```r
# Teste inicial
1 + 1

# Corpus canônico de 8 documentos da disciplina
docs <- c(
  d1 = "recuperacao de informacao ordena documentos por relevancia",
  d2 = "o modelo de espaco vetorial representa documentos como vetores",
  d3 = "bm25 e um modelo probabilistico de ranqueamento de texto",
  d4 = "aprendizado estatistico fundamenta a recuperacao moderna",
  d5 = "o indice invertido acelera a busca em muitos documentos",
  d6 = "embeddings capturam a semantica de palavras e documentos",
  d7 = "a avaliacao mede a relevancia dos resultados da busca",
  d8 = "ciencia de dados combina estatistica e programacao"
)

length(docs)   # quantos elementos tem o vetor?
docs[1]        # primeiro elemento (R comeca em 1)
docs[0]        # no R devolve vetor vazio (named character(0)), sem erro

# Operacoes vetorizadas
c(1, 2, 3) * 2

# Checkpoint 1
length(docs[c(2, 4, 6)])
c(10, 20) * 3
```

## Módulo 2 — Nomes: `[ ]`, `[[ ]]`, `names`
Diferença entre preservar o nome (`[ ]`) e extrair apenas o valor (`[[ ]]`).

```r
docs["d5"]      # com o nome em cima
docs[["d5"]]    # apenas o valor
names(docs)     # vetor com todos os nomes

# Comparacao e nome grudado
docs[["d1"]] == docs[1]

# Checkpoint 2
names(docs)[c(1, 8)]
docs[["d2"]] # colchete duplo tira o nome
```

## Módulo 3 — Texto I: `nchar`, `toupper`, `tolower`, `substr`, `paste`
Comprimento de strings vs comprimento de vetores e manipulação básica de texto.

```r
nchar(docs)                    # caracteres de cada documento
toupper(docs[["d8"]])          # maiusculas
substr(docs[["d1"]], 1, 11)    # do 1 ao 11 inclusive

paste0("d", 1:8)                          # sem separador
paste("doc", names(docs))                 # com espaco
paste(c("a", "b", "c"), collapse = "-")   # junta vetor

# Checkpoint 3
nchar(c("bm25", "de"))
length(c("bm25", "de"))
```

## Módulo 4 — Texto II: `strsplit` e `unlist`
Quebra de texto em listas e achatamento em vetores simples.

```r
frase <- substr(docs[["d1"]], 1, 25)
strsplit(frase, " ")         # devolve lista
unlist(strsplit(frase, " ")) # vira vetor

# Checkpoint 4
length(unlist(strsplit("bm25 e um", " ")))
typeof(strsplit("bm25 e um", " "))
```

## Módulo 5 — Funções: `tokenizar` e `maiuscula`
Criação de funções em R; retorno da última expressão avaliada.

```r
tokenizar <- function(texto) {
  texto <- tolower(texto)
  unlist(strsplit(texto, " "))
}
tokenizar(frase)
tokenizar("Ciencia de DADOS")

dobro <- function(x) x * 2
dobro(c(1, 2, 3))

# Checkpoint 5: maiuscula
maiuscula <- function(texto) toupper(unlist(strsplit(texto, " "))[1])
maiuscula(docs[["d3"]])
```

## Módulo 6 — `lapply` e `sapply`
Aplicação funcional sobre coleções: `lapply` preserva listas, `sapply` simplifica.

```r
tokens <- lapply(docs, tokenizar)
length(tokens)
tokens[["d4"]]

sapply(tokens, length)
sum(sapply(tokens, length)) # total de palavras no corpus

# Checkpoint 6
sapply(list(a = 1:3, b = 1:5), sum)
```

## Módulo 7 — `table`, `factor`, `sort`, `%in%`
Contagem de termos, fixação de categorias com zeros via `factor` e filtragem com `%in%`.

```r
table(tokens[["d3"]])["de"]
sort(table(tokens[["d3"]]), decreasing = TRUE)[1:3]

vocab <- c("de", "modelo", "busca")
table(factor(tokens[["d3"]], levels = vocab))

c("modelo", "busca") %in% tokens[["d3"]]
tokens[["d1"]][!tokens[["d1"]] %in% c("de", "por")]

# Checkpoint 7
table(factor(tokens[["d2"]], levels = vocab))
```

## Módulo 8 — Matrizes e reciclagem
Preenchimento por coluna e multiplicação com reciclagem matricial.

```r
m <- matrix(c(1, 0, 1, 1), nrow = 2)
rownames(m) <- c("de", "modelo")
colnames(m) <- c("d1", "d2")
m

peso <- c(1, 10)
m * peso

# Checkpoint 8
m * c(2, 3)
```

## Módulo 9 — Regex I — Procurar: `grep`, `grepl`, `^`, `$`, `|`, `[ ]`, `+`
Busca por padrões, âncoras e a lição fundamental: 'pedaço não é palavra'.

```r
grep("modelo", docs)
grepl("modelo", docs)
names(docs)[grepl("modelo", docs)]

grep("^o ", docs)
grep("busca$", docs)
grep("busca|modelo", docs)
grep("[0-9]", docs)

# Pedaço não é palavra (d4 e d7 entram porque contêm 'moderna' e 'mede')
grep("de", docs)
grep(" de ", docs)

# Checkpoint 9
grep("documentos$", docs)
grep("^e", docs)
```

## Módulo 10 — Regex II — Transformar e limpar: `sub`, `gsub`, `.`, `*`, `{n}`, `[^ ]`, `trimws`
Substituição pontual vs global, faxina de texto em 4 linhas e tokenização robusta com `\\s+`.

```r
sub("de", "DE", docs[["d3"]])
gsub("de", "DE", docs[["d3"]])

gsub("[aeiou]", "", docs[["d1"]])
gsub("[0-9]+", "", docs[["d3"]])
grep("[a-z]{12}", docs)

sub(" .*$", "", docs[["d1"]])
sub("^.* ", "", docs[["d1"]])

# Pipeline de higienizacao em 4 linhas
sujo <- "  O Modelo   de Espaco   Vetorial!! "
x_limpo <- trimws(sujo)
x_limpo <- tolower(x_limpo)
x_limpo <- gsub("[^a-z ]", "", x_limpo)
x_limpo <- gsub("\\s+", " ", x_limpo)
x_limpo

# Função tokenizar definitiva (compatível com múltiplos espaços)
tokenizar <- function(texto) {
  unlist(strsplit(tolower(texto), "\\s+"))
}

# Checkpoint 10
gsub("o", "0", docs[["d5"]])
sub(" .*$", "", docs[["d3"]])
```
