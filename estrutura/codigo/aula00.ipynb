## 1. Setup e Corpus Canônico (Módulos 1–11)
Definição do vetor canônico `docs` e da função definitiva `tokenizar`.

```r
# 1. Corpus canônico de 8 documentos da disciplina
docs <- c(
  d1 = "recuperacao de informacao ordena documentos por relevancia",
  d2 = "o modelo de espaco vetorial representa documentos como vetores",
  d3 = "bm25 e um modelo probabilistico de ranqueamento de texto",
  d4 = "aprendizado estatistico fundamenta a recuperacao moderna",
  d5 = "o indice invertido acelera a busca em muitos documentos",
  d6 = "embeddings capturam a semantica de palavras e documentos",
  d7 = "a avaliacao mede a relevancia dos resultados da busca",
  d8 = "ciencia de dados combina estatistica e programacao"
)

# 2. Função de tokenização definitiva (suporta espaços múltiplos)
tokenizar <- function(texto) {
  unlist(strsplit(tolower(texto), "\\s+"))
}

cat("Setup concluido: docs com", length(docs), "documentos e tokenizar() pronta.\n")
```

## 2. Módulo 12 — Três frases suas: sujar, limpar, contar
Criação de 3 frases deliberadamente sujas, aplicação do pipeline de 4 linhas e contagem.

```r
# 1. Três frases sujas (espaços nas pontas e meio, maiúsculas/minúsculas, números e pontuação)
f <- c(
  f1 = "  O Modelo de dados   apresenta Alta complexidade! ",
  f2 = "A ciencia de DADOS moderna exige 2026 modelos de aprendizado. ",
  f3 = "   modelos estatisticos fundamentam a busca moderna? "
)

# 2. Pipeline de limpeza em 4 linhas
x <- trimws(f)               # 1. Remove espaços das extremidades
x <- tolower(x)              # 2. Converte para minúsculas
x <- gsub("[^a-z ]", "", x)  # 3. Remove tudo exceto letras e espaços
x <- gsub("\\s+", " ", x)    # 4. Reduz espaços múltiplos a um só

print(x)
```

```r
# 3. Tokenização e frequência
tokens <- lapply(x, tokenizar)
freq   <- table(unlist(tokens))
print(sort(freq, decreasing = TRUE))

# 4. Inspeção de caracteres e vocabulário
cat("\nCaracteres antes da limpeza:\n")
print(nchar(f))
cat("\nCaracteres apos a limpeza:\n")
print(nchar(x))

cat("\nTotal de palavras:", length(unlist(tokens)), "\n")
cat("Termos unicos:", length(unique(unlist(tokens))), "\n")
```

## 3. Módulo 13 — Regex nas suas frases
Validação de padrões nas frases originais `f` e nas frases limpas `x`.

```r
# 1. Começam com maiúscula (ignorando espaços iniciais)
cat("1. Comecam com maiuscula:\n")
print(grepl("^\\s*[A-Z]", f))

# 2. Contêm dígito/número
cat("\n2. Contem numeros:\n")
print(grepl("[0-9]", f))

# 3. Terminam em pontuação (após trimws)
cat("\n3. Terminam em pontuacao:\n")
print(grepl("[^a-z ]$", trimws(f)))

# 4. Primeira palavra de cada frase limpa
cat("\n4. Primeira palavra de cada frase:\n")
print(sub(" .*$", "", x))

# 5. Troca de espaços por sublinhado (_)
cat("\n5. Espacos trocados por _:\n")
print(gsub(" ", "_", x))

# Checkpoint 13: Começam com maiúscula e terminam em 'a' (antes da pontuação)
cat("\nCheckpoint 13 (Comeca maiuscula e termina em 'a'):\n")
print(grepl("^\\s*[A-Z].*a[^a-z]*$", trimws(f)))
```
